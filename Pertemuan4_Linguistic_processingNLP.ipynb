```python
import spacy
import pandas as pd
from spacy import displacy

# Download model dulu
import subprocess
subprocess.run(["python", "-m", "spacy", "download", "xx_ent_wiki_sm"])
```

```python
# Load model yang sudah terdownload
nlp = spacy.load("xx_ent_wiki_sm")
print("Model berhasil dimuat")
```

```python
doc = nlp("Mahasiswa belajar Natural Language Processing.")

print(f"{'Token':<15} {'POS':<10} {'Dependency'}")
print("-" * 45)
for token in doc:
    print(f"{token.text:<15} {token.pos_:<10} {token.dep_}")
```

```python
doc = (nlp("Mahasiswa belajar Natural Language Processing."))
print(f"{'Token':<15} {'POS':<10} {'Lemma'}")
print("-" * 45)
for token in doc:
    print(f"{token.text:<15} {token.pos_:<10} {token.lemma_}")
```

```python
import pandas as pd

def analyze_pos(text, nlp_model):
    doc = nlp_model(text)
    results = []
    for token in doc:
        results.append({
            "Token": token.text,
            "POS": token.pos_,
            "POS Tag": token.tag_,
            "Lemma": token.lemma_,
            "Dependency": token.dep_,
            "Head": token.head.text
        })
    return pd.DataFrame(results)

df_pos = analyze_pos("Mahasiswa belajar NLP.", nlp)
print(df_pos.to_string(index=False))
```

```python
sentences = [
    "Mahasiswa belajar NLP.",
    "Dosen mengajar Natural Language Processing.",
    "Mahasiswa membaca buku.",
    "Sistem memproses data teks.",
    "Model bahasa memprediksi kata berikutnya."
]

for sentence in sentences:
    print(f"\nKalimat: {sentence}")
    df_result = analyze_pos(sentence, nlp)
    print(df_result[["Token", "POS", "Dependency", "Head"]].to_string(index=False))
    print()
```

```python
# === BAGIAN 9: ANALISIS DATASET ===

import pandas as pd

# Baca dataset
df = pd.read_csv("news_data.csv")

text_column = 'article_text'

# Ambil sampel 5 kalimat
sample = df.sample(5, random_state=42)

print("="*60)
print("HASIL ANALISIS DISTRIBUSI KELAS KATA")
print("="*60)

for idx, row in sample.iterrows():
    teks = str(row[text_column])[:150]
    doc = nlp(teks)
    
    pos_counts = {}
    for token in doc:
        pos = token.pos_
        if pos:
            pos_counts[pos] = pos_counts.get(pos, 0) + 1
    
    print(f"\nTeks: {teks}...")
    print(f"Distribusi POS: {pos_counts}")

print("\nSelesai!")
```

```python
# === BAGIAN 10: ANALISIS KESALAHAN ===

teks_informal = [
    "aku lagi belajar NLP nih",
    "gue udah baca buku itu",
    "modelnya bagus banget sih",
    "nggak ngerti sama sekali",
    "udah instal belum?"
]

print("="*60)
print("ANALISIS TEKS INFORMAL & KETERBATASAN MODEL")
print("="*60)

for teks in teks_informal:
    doc = nlp(teks)
    print(f"\nTeks: {teks}")
    print("-" * 40)
    for token in doc:
        label = token.pos_ if token.pos_ else "(tidak terdeteksi)"
        print(f"  {token.text:<15} → {label}")

print("\nSelesai!")
```
