```python
# Pastikan library sudah diimpor
import pandas as pd
from nltk.tokenize import word_tokenize

# 1. Baca dataset
df = pd.read_csv("news_data.csv")

# 2. Cek nama kolom
print("Kolom tersedia:", df.columns.tolist())

# 3. Tentukan kolom teks
text_column = "article_text"

# 4. Buat fungsi tokenisasi
def tokenisasi(teks):
    return word_tokenize(str(teks))

# 5. Terapkan ke seluruh dataset
df["tokens"] = df[text_column].apply(tokenisasi)

# 6. Tampilkan hasil
print("\n Hasil Tokenisasi:")
print(df[[text_column, "tokens"]].head())

# 7. Hitung jumlah token
df["jumlah_token"] = df["tokens"].apply(len)
print("\n Ringkasan Jumlah Token:")
print(df["jumlah_token"].describe())
```
